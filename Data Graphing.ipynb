```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
d = [[4, 9, 13, 25, 29, 44, 56, 69, 116], [3, 8, 18, 26, 38, 64, 109, 161, 218], [None, 10, 15, 36, 51, 59, 75, 127, 142]]

df = pd.DataFrame()
df['No LD'] = d[0]
df['1 LD'] = d[1]
df['2 LD'] = d[2]

# df = df.reindex(range(2, 11))

print(df)

fig, ax = plt.subplots(figsize=(6, 5))

sns.lineplot(data=df, ax=ax, color='orange', marker='s', legend=True)
ax.set(xlabel='Dimension', ylabel='Execution Time (μs)')
ax.set_xticks(range(len(df)), labels=range(2, 11))
```
