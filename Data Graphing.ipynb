```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
d = [[2, 4, 8, 12, 18, 23, 32, 41, 55], [1, 1, 2, 3, 4, 6, 7, 9, 11], [2, 4, 8, 14, 21, 31, 38, 52, 72]]

df = pd.DataFrame()
df['Random'] = d[0]
df['Standard'] = d[1]
df['Random LLL'] = d[2]

print(df)

fig, ax = plt.subplots(figsize=(6, 5))

sns.lineplot(data=df, ax=ax, marker='s', legend=True)
ax.set(xlabel='Dimension', ylabel='Execution Time (μs)')
ax.set_xticks(range(len(df)), labels=range(2, 11))
```

```python
d = [19, 33, 50, 70, 93, 119, 148, 180, 215]

fig, ax = plt.subplots(figsize=(6, 5))

sns.lineplot(data=d, ax=ax, color='orange', marker='s', legend=True)
ax.set(xlabel='Dimension', ylabel='Amount of Dimension Affected Data')
ax.set_xticks(range(len(d)), labels=range(2, 11))
```
